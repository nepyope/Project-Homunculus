```python
#include <Arduino.h>
#include <Wire.h>
#include <Adafruit_PWMServoDriver.h>

// Define GPIO pins for analog and button inputs
const int analogPins[] = {36, 39, 35, 34, 33, 32, 25, 26};
const int numAnalogPins = sizeof(analogPins) / sizeof(analogPins[0]);
int analogValues[numAnalogPins];

const int buttonPins[] = {5, 17};
const int numButtonPins = sizeof(buttonPins) / sizeof(buttonPins[0]);
int buttonValues[numButtonPins];

const int defaultPin = 16;

// Servo control definitions
Adafruit_PWMServoDriver pwm = Adafruit_PWMServoDriver();
#define SERVOMIN  150 // Minimum pulse length out of 4096
#define SERVOMAX  600 // Maximum pulse length out of 4096
#define SERVO_FREQ 50 // Analog servos run at ~50 Hz updates

void setup() {
  Serial.begin(115200);
  delay(1000);

  // Initialize GPIO pins
  for (int i = 0; i < numAnalogPins; i++) {
    pinMode(analogPins[i], INPUT);
  }
  for (int i = 0; i < numButtonPins; i++) {
    pinMode(buttonPins[i], INPUT_PULLUP);
  }
  pinMode(defaultPin, OUTPUT);
  digitalWrite(defaultPin, HIGH);

  // Initialize servo control
  pwm.begin();
  pwm.setOscillatorFrequency(27000000);
  pwm.setPWMFreq(SERVO_FREQ);
  delay(10);
}

void loop() {
  // Read analog values
  for (int i = 0; i < numAnalogPins; i++) {
    analogValues[i] = analogRead(analogPins[i]);
  }

  // Read button values and flip the last bit for the second button
  buttonValues[0] = digitalRead(buttonPins[0]);
  buttonValues[1] = digitalRead(buttonPins[1]) ^ 1;

  // Print values
  for (int i = 0; i < numAnalogPins; i++) {
    Serial.print(analogValues[i]);
    Serial.print(",");
  }
  for (int i = 0; i < numButtonPins; i++) {
    Serial.print(buttonValues[i]);
    if (i < numButtonPins - 1) {
      Serial.print(",");
    }
  }
  Serial.println();

  // Servo control based on serial input
  if (Serial.available() > 0) {
    int servoNum = Serial.parseInt();
    int angle = Serial.parseInt();

    if (Serial.read() == '\n') {
      int pulse = map(angle, 0, 180, SERVOMIN, SERVOMAX);
      pwm.setPWM(servoNum, 0, pulse);
    }
  }

  delay(10);
}
```

```python
import pygame
import serial
import time

# Initialize Pygame
pygame.init()
screen = pygame.display.set_mode((800, 600))
pygame.display.set_caption("Servo and Sensor Visualizer")

# Colors and Fonts
WHITE = (255, 255, 255)
BLACK = (0, 0, 0)
GREEN = (0, 255, 0)
FONT = pygame.font.SysFont('Arial', 20)

# Serial Communication
ser = serial.Serial('/dev/ttyUSB1', 115200)

def set_servo(servo_num, angle):
    command = f"{servo_num} {angle}\n"
    ser.write(command.encode())

class MovingAverage:
    def __init__(self, window_size):
        self.window_size = window_size
        self.values = []

    def add_value(self, value):
        self.values.append(value)
        if len(self.values) > self.window_size:
            self.values.pop(0)
        return self.average()

    def average(self):
        if self.values:
            return sum(self.values) / len(self.values)
        return 0

# Initialize filters for each sensor
sensor_filters = [MovingAverage(10) for _ in range(8)]


# Initial servo values
servo_values = [55 + i for i in range(4, 12)]
sensor_values = [0 for _ in range(8)]
s_values = [0 for _ in range(8)]
button_values = [0, 0]  # Assuming two buttons

def map_value(value, min_in, max_in, min_out, max_out):
    # Map a value from one range to another
    return (value - min_in) * (max_out - min_out) / (max_in - min_in) + min_out

def get_angle_from_click(y_pos):
    # Map the y-coordinate to the angle range (20 to 90 degrees)
    return int(map_value(y_pos, 100, 0, 20, 90))

def get_sensor_display_height(sensor_value):
    # Map sensor value to the display height (1800 to 3800 mapped to 0 to 100)
    return int(map_value(sensor_value, 1800, 3800, 100, 0))

# Main loop
running = True
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        elif event.type == pygame.MOUSEBUTTONDOWN:
            x, y = event.pos
            if 300 <= y <= 400:  # Y coordinate within the boxes
                box_index = x // 100  # Determine which box was clicked
                if 0 <= box_index < 8:
                    relative_y = y - 300  # Y position relative to the top of the box
                    servo_angle = get_angle_from_click(relative_y)
                    servo_values[box_index] = servo_angle
                    set_servo(box_index + 4, servo_angle)

    # Read data from Arduino
    if ser.in_waiting > 0:
        line = ser.readline().decode('utf-8').rstrip()
        values = line.split(',')
        if len(values) != 10:
            continue
        sensor_values = list(map(int, values[:8]))
        for i in range(8):
            sensor_values[i] = sensor_filters[i].add_value(sensor_values[i])
        s_values = [sensor_values[6], sensor_values[3], sensor_values[2], sensor_values[4], sensor_values[7], sensor_values[5], sensor_values[0], sensor_values[1]]
        button_values = list(map(int, values[8:10]))

    # Drawing
    screen.fill(BLACK)
    for i in range(8):
        # Draw box for each servo and sensor
        box_rect = pygame.Rect(i * 100, 300, 80, 100)
        pygame.draw.rect(screen, WHITE, box_rect, 2)

        # Fill left half of the box based on servo angle
        angle_height = map_value(servo_values[i], 20, 90, 100, 0)
        servo_rect = pygame.Rect(i * 100, 300, 40, angle_height)
        pygame.draw.rect(screen, WHITE, servo_rect)

        # Fill right half of the box based on sensor value
        sensor_height = get_sensor_display_height(s_values[i])
        sensor_rect = pygame.Rect(i * 100 + 40, 300, 40, sensor_height)
        pygame.draw.rect(screen, GREEN, sensor_rect)

        # Display servo angle and sensor reading
        angle_text = FONT.render(f"Angle: {servo_values[i]}", True, WHITE)
        sensor_text = FONT.render(f"{s_values[i]}", True, WHITE)
        screen.blit(angle_text, (i * 100 + 10, 260))
        screen.blit(sensor_text, (i * 100 + 10, 410))

        # Display button state boxes
        button_color = WHITE if button_values[i % 2] else BLACK
        pygame.draw.rect(screen, button_color, (i * 100 + 10, 10, 20, 20))

    pygame.display.flip()
    time.sleep(0.001)
```

```python
import pygame
import sys
import math
import numpy as np
# Initialize Pygame
pygame.init()

# Screen dimensions and background color
screen_width, screen_height = 800, 600
background_color = (50, 205, 50)  # Lime color

# Pole properties
pole_color = (139, 69, 19)  # Brown color
pole_length = 200  # Length of the pole
fixed_x, fixed_y = 400, 300  # Fixed bottom position of the pole
pole_mass = 1  # Mass of the pole (adjustable)

# Spring properties0
spring_color = (128, 128, 128)  # Grey color
spring_thickness = 2  # Thickness of the spring line
spring_rest_length = 50  # Rest length of the spring
spring_constant = 1  # Spring constant

# Force arrow properties
force_arrow_thickness = 5
spring_force_color = (0, 0, 255)  # Blue color for spring force
gravity_force_color = (255, 165, 0)  # Orange color for gravity force
arrowhead_length = 10  # Length of the arrowhead

# Point properties
point_radius = 250  # Initial radius larger than the pole
point_color = (255, 0, 0)  # Red color

# Gravity and rotation
gravity = 0.001
angle = 0  # Initial angle in radians
angular_velocity = 0
angular_acceleration = 0
# Moment of Inertia (initial value)
moment_of_inertia = (1/3) * pole_mass * pole_length**2  # Simplified for this example


# Create the screen
screen = pygame.display.set_mode((screen_width, screen_height))

# Game loop
while True:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            pygame.quit()
            sys.exit()

    # Get mouse position
    mouse_x, mouse_y = pygame.mouse.get_pos()
    
    # Calculate the angle between the center and mouse cursor
    mouse_angle = math.atan2(mouse_y - fixed_y, mouse_x - fixed_x)
    
    # Calculate the position of the point at the radius from the center
    point_x = fixed_x + point_radius * math.cos(mouse_angle)
    point_y = fixed_y + point_radius * math.sin(mouse_angle)
    
    # Calculate the top position of the pole
    top_x = fixed_x + pole_length * math.sin(angle)
    top_y = fixed_y - pole_length * math.cos(angle)

    # Calculate spring force
    spring_length = math.sqrt((top_x - point_x)**2 + (top_y - point_y)**2)
    spring_extension = spring_length - spring_rest_length
    spring_force_magnitude = spring_constant * spring_extension
    spring_force_angle = math.atan2(point_y - top_y, point_x - top_x)+np.pi/2

    spring_torque = spring_force_magnitude * pole_length * math.sin(spring_force_angle - angle)

    # Calculate the torque due to gravity (acting at the center of mass)
    gravity_torque = gravity * (pole_length / 2) * math.sin(angle)

    # Total torque is the sum of spring torque and gravity torque
    total_torque = spring_torque - gravity_torque

    # Update moment of inertia based on current pole mass
    moment_of_inertia = (1/3) * pole_mass * pole_length**2
    
    # Update angular acceleration, velocity, and angle
    angular_acceleration = total_torque / moment_of_inertia
    angular_velocity += angular_acceleration
    angle += angular_velocity

    # Clear the screen
    screen.fill(background_color)

    # Draw the pole
    pygame.draw.line(screen, pole_color, (fixed_x, fixed_y), (top_x, top_y), 5)

    # Draw line representing the spring's force
    force_line_length = spring_force_magnitude * 100  # Scale for visibility
    force_line_end_x = top_x + force_line_length * math.cos(spring_force_angle)
    force_line_end_y = top_y + force_line_length * math.sin(spring_force_angle)
    pygame.draw.line(screen, spring_force_color, (top_x, top_y), (force_line_end_x, force_line_end_y), force_arrow_thickness)
    
    # Draw the spring
    pygame.draw.line(screen, spring_color, (top_x, top_y), (int(point_x), int(point_y)), spring_thickness)
    
    # Draw the controlled point
    pygame.draw.circle(screen, point_color, (int(point_x), int(point_y)), 5)
    
    # Update the display
    pygame.display.flip()

    # Frame rate
    pygame.time.Clock().tick(10)
```

```python
import pygame
import math

# Initialize Pygame
pygame.init()

# Screen dimensions and background color
WIDTH, HEIGHT = 800, 600
BACKGROUND_COLOR = (255, 255, 255)  # White

# Display setup
screen = pygame.display.set_mode((WIDTH, HEIGHT))
pygame.display.set_caption("Interactive Rotating Sticks Simulation")

# Constants
L = 100  # Length of the sticks
D = 150  # Distance between the first set of sticks

# Stick angles and drag states
angle1 = 0
angle2 = 0
dragging1 = False
dragging2 = False

# Function to draw a stick
def draw_stick(center_x, center_y, length, angle):
    end_x = center_x + length * math.cos(angle)
    end_y = center_y + length * math.sin(angle)
    pygame.draw.line(screen, (0, 0, 0), (center_x, center_y), (end_x, end_y), 2)
    return end_x, end_y

# Function to calculate the intersection of two circles
def circle_intersection(circle1, circle2):
    x1, y1, r1 = circle1
    x2, y2, r2 = circle2
    dx, dy = x2 - x1, y2 - y1
    d = math.sqrt(dx*dx + dy*dy)
    if d > r1 + r2:
        return None  # No solutions, the circles are separate
    if d < abs(r1 - r2):
        return None  # No solutions, one circle is contained within the other
    if d == 0 and r1 == r2:
        return None  # Circles are coincident and there are an infinite number of solutions

    a = (r1*r1 - r2*r2 + d*d) / (2*d)
    h = math.sqrt(r1*r1 - a*a)
    xm = x1 + a*dx / d
    ym = y1 + a*dy / d
    xs1 = xm + h*dy / d
    xs2 = xm - h*dy / d
    ys1 = ym - h*dx / d
    ys2 = ym + h*dx / d

    return (xs1, ys1), (xs2, ys2)
# Function to calculate angle based on mouse position
def calculate_angle(center_x, center_y, mouse_x, mouse_y):
    return math.atan2(mouse_y - center_y, mouse_x - center_x)

# Function for handling stick dragging
def handle_dragging(event, stick_num):
    global angle1, angle2, dragging1, dragging2
    mouse_x, mouse_y = event.pos

    if stick_num == 1:
        center_x, center_y = WIDTH / 2 - D / 2, HEIGHT / 2
        dragging1 = True
    else:
        center_x, center_y = WIDTH / 2 + D / 2, HEIGHT / 2
        dragging2 = True

    if dragging1 or dragging2:
        angle = calculate_angle(center_x, center_y, mouse_x, mouse_y)
        if stick_num == 1:
            angle1 = angle
        else:
            angle2 = angle

# Main loop
running = True
clock = pygame.time.Clock()
while running:
    for event in pygame.event.get():
        if event.type == pygame.QUIT:
            running = False
        if event.type == pygame.MOUSEBUTTONDOWN:
            if event.button == 1:  # Left mouse button
                # Check which stick is closer to the mouse click
                mouse_x, mouse_y = event.pos
                center_x1, center_y1 = WIDTH / 2 - D / 2, HEIGHT / 2
                center_x2, center_y2 = WIDTH / 2 + D / 2, HEIGHT / 2
                distance1 = math.sqrt((mouse_x - center_x1)**2 + (mouse_y - center_y1)**2)
                distance2 = math.sqrt((mouse_x - center_x2)**2 + (mouse_y - center_y2)**2)
                if distance1 < distance2:
                    handle_dragging(event, 1)
                else:
                    handle_dragging(event, 2)
        if event.type == pygame.MOUSEBUTTONUP:
            if event.button == 1:  # Left mouse button
                dragging1 = dragging2 = False
        if event.type == pygame.MOUSEMOTION:
            if dragging1 or dragging2:
                handle_dragging(event, 1 if dragging1 else 2)

    # Fill the screen
    screen.fill(BACKGROUND_COLOR)

    # Draw first set of sticks and get their tips
    tip1 = draw_stick(WIDTH / 2 - D / 2, HEIGHT / 2, L, angle1)
    tip2 = draw_stick(WIDTH / 2 + D / 2, HEIGHT / 2, L, angle2)

    # Calculate intersection points for the second set of sticks
    intersections = circle_intersection((*tip1, L), (*tip2, L))
    if intersections:
        # Choose one of the intersection points
        intersection = intersections[0]

        # Draw second set of sticks
        draw_stick(*tip1, L, math.atan2(intersection[1]-tip1[1], intersection[0]-tip1[0]))
        draw_stick(*tip2, L, math.atan2(intersection[1]-tip2[1], intersection[0]-tip2[0]))

    # Update the display
    pygame.display.flip()

    # Cap the frame rate
    clock.tick(60)

pygame.quit()
```
